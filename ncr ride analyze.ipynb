```python
# Booking Status anlyze
booking_status = df["Booking Status"].value_counts()
print(booking_status)
```

```python
# visualization
booking_status.plot(kind="bar")
plt.title("Booking Status")
plt.xlabel("Status")
plt.ylabel("Value")
plt.show()
```

```python
# vehicle type ananlyze
vehicle_type = df["Vehicle Type"].value_counts()
print(vehicle_type)

# visualization
vehicle_type.plot(kind="bar")
plt.title("Vehicle Type Analyze")
plt.xlabel("Vehicles")
plt.ylabel("Value count")
plt.show()
```

```python
# Pickup location Analyze
top_pickup_loc = df["Pickup Location"].value_counts().head(10)
print(top_pickup_loc)
```

```python
# Top Drop Location
top_drop_location = df["Drop Location"].value_counts().head(10)
print(top_drop_location)
```

```python
# reason for cancelling ride by customers
cancelled_ride_by_customers = df["Reason for cancelling by Customer"].value_counts()
print(cancelled_ride_by_customers)
```

```python
# Reason for driver cancellation
cancelled_by_driver = df["Driver Cancellation Reason"].value_counts()
print(cancelled_by_driver)
```

```python
# reson for incomplete ride reason
incomplete_ride = df["Incomplete Rides Reason"].value_counts()
print(incomplete_ride)
```

```python
# total revenue
total_revenue = df["Booking Value"].sum()
print("Total Revenue :",total_revenue)
```

```python
# average ride distance
ride_distance = df["Ride Distance"].mean()
print("average ride distance :",ride_distance)
```

```python
# customers rating analyze
df["Customer Rating"].describe()
```

```python
# payment method analyze
payment_method = df["Payment Method"].value_counts()
print(payment_method)

# visualization
payment_method.plot(kind="pie", autopct = "%1.1f%%")
plt.title("Payment Method analyze")
plt.ylabel("")
plt.show()
```

```python
# total revenue
total_revenue = df["Booking Value"].sum()
print("Total Revenue :",total_revenue)
```

```python
# vehicle wise revenue
vehicle_wise_revenue = df.groupby("Vehicle Type")["Booking Value"].sum()
print(vehicle_wise_revenue.sort_values(ascending = False))
```

```python
# location wise revenue
loc_wise_revn = df.groupby("Drop Location")["Booking Value"].sum().head(10)
print(loc_wise_revn.sort_values(ascending=False))
```

```python
# average fare
average_fare = df["Booking Value"].mean()
print(average_fare)

avg = df.groupby("Vehicle Type")["Booking Value"].mean()
print(avg)
```

```python
# booking status wise revenue
revenue_by_booking = df.groupby("Booking Status")["Booking Value"].sum()
print(revenue_by_booking)
```

```python
# highest revenue vehicle
high_revn_vehicle = df.groupby("Vehicle Type")["Booking Value"].sum()
print(high_revn_vehicle.sort_values(ascending=False))
```

```python
# Monthly revenue trend
df["Date"] = pd.to_datetime(df["Date"])
monthly_revenue = df.groupby(df["Date"].dt.month)["Booking Value"].sum()
print(monthly_revenue)
```

```python
# vehicle type wise rating
vehicle_wise_Rating = df.groupby("Vehicle Type")["Customer Rating"].mean()
print(vehicle_wise_Rating)
```

```python
# Time Analysis
df["Time"] = pd.to_datetime(df["Time"]).copy()
booking_by_time = df.groupby(df["Time"].dt.hour)["Booking ID"].count()
print(booking_by_time)
```

```python
# vehicle wise payment method
vehicle_wise_payment = df.groupby("Vehicle Type")["Payment Method"].value_counts()
print(vehicle_wise_payment)
```

```python
# vehicle wise booking
total_booking = df.groupby("Vehicle Type")["Booking ID"].count()
print(total_booking)
```

```python
# Vehicle-wise Cancellation Rate
cancelled = df[
    df["Booking Status"].str.contains(
        "Cancelled",
        case=False,
        na=False
    )
]

cancelled_bookings = cancelled.groupby(
    "Vehicle Type"
)["Booking ID"].count()
```

```python
cancellation_rate = (
    cancelled_bookings / total_booking
) * 100

print(cancellation_rate)
```

```python
result = pd.DataFrame({
    "Total Bookings": total_booking,
    "Cancelled Bookings": cancelled_bookings
})

result["Cancellation Rate (%)"] = (
    result["Cancelled Bookings"]
    / result["Total Bookings"]
) * 100

print(result.fillna(0))
```
